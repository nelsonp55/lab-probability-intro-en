# Probability

## Challenge 1
There is a box with 10 white balls, 12 red balls and 8 black balls. Calculate the probability of:
* Taking a white ball out.
* Taking a white ball on first extraction and a black ball out on second extraction without replacing the first one.
* Taking a white ball on first extraction and a black ball out on second extraction replacing the first one.
* Taking 3 balls and them being all different colours.

```python
white = 10
red = 12
black = 8

total = white + red + black

# 1
p_white = white / total

# 2 - without replacement
p_without = (white / total) * (black / (total - 1))

# 3 - with replacement
p_with = (white / total) * (black / total)

# 4 - all different colours
p_diff = 6 * (white / total) * (red / (total - 1)) * (black / (total - 2))

print("1. Probability of taking a white ball:")
print(p_white)

print("\n2. White then black without replacement:")
print(p_without)

print("\n3. White then black with replacement:")
print(p_with)

print("\n4. Three balls all different colours:")
print(p_diff)
```

## Challenge 2

You want to go on picnic on Saturday. You hate rain, and would know to plan your weekend. To help you make a decision, you gather the following data about rainy days:

* 40% of all days are cloudy.
* About 50% of cloudy days are rainy. 
* About 20% of non-cloudy days are rainy.

What is the chance of rain during next Saturday?

```python
p_train = 0.60
p_drive = 0.40

p_late_train = 0.05
p_late_drive = 0.20

p_late = (p_late_train * p_train) + (p_late_drive * p_drive)

print(f"Probability of being late: {p_late:.2%}")
```

## Challenge 3

A and B are events of a probability such that $P(A) = 0.3$, $P(B) = 0.6$ and $P(A \cap B) = 0.1$

Which of the following statements are true?  
**HINT: draw a big Venn Diagram by your side**  
* $P(\overline{A}) = 0.7$
* $P(A \cup B) = 0.6$
* $P(A \cap \overline{B}) = 0.2$
* $P(A \cup \overline{A}) = 1$  
* $P(A \cap (B \cup \overline{B})) = 0.4$
* $P(\overline{A} \cap \overline{B}) = 0.3$
* $P(\overline{A \cap B}) = 0.9$

```python
P_A = 0.3
P_B = 0.6
P_A_intersection_B = 0.1

# 1
print("1.", 1 - P_A == 0.7)

# 2
P_union = P_A + P_B - P_A_intersection_B
print("2.", P_union == 0.6)

# 3
P_A_and_not_B = P_A - P_A_intersection_B
print("3.", P_A_and_not_B == 0.2)

# 4
print("4.", 1 == 1)

# 5
print("5.", P_A == 0.4)

# 6
P_not_A_and_not_B = 1 - P_union
print("6.", P_not_A_and_not_B == 0.3)

# 7
print("7.", 1 - P_A_intersection_B == 0.9)
```

## Challenge 4

 * If someone flies to China, the probability of getting Covid is 1%  
 * If someone flies to Italy, the probability of getting Covid is 20%  

I roll 1 dice with numbers 1, 2, 3, 4, 5, 6:  
 * If 1: I fly to Italy
 * If not 1: I fly to China

During my trip, I got Covid. What is the probability that I flew to China?

```python
p_china = 5/6
p_italy = 1/6

p_covid_china = 0.01
p_covid_italy = 0.20

# Total probability of getting covid
p_covid = (p_covid_china * p_china) + (p_covid_italy * p_italy)

# Bayes theorem
p_china_given_covid = (p_covid_china * p_china) / p_covid

print(f"Probability of having flown to China given Covid: {p_china_given_covid:.4f}")
print(f"Percentage: {p_china_given_covid:.2%}")
```

## BONUS

You roll 2 dice. What is the probability that the sum of the numbers is 7? And 8? And 9? Then simulate 10000 times and draw a countplot.

your explanation here

```python
import random
import pandas as pd
import matplotlib.pyplot as plt

results = []

# simulate 10000 rolls
for _ in range(10000):

    dice1 = random.randint(1, 6)
    dice2 = random.randint(1, 6)

    total = dice1 + dice2

    results.append(total)

# dataframe
df = pd.DataFrame(results, columns=["sum"])

# probabilities
print("P(7) =", 6/36)
print("P(8) =", 5/36)
print("P(9) =", 4/36)

# countplot
df["sum"].value_counts().sort_index().plot(kind="bar")

plt.xlabel("Sum")
plt.ylabel("Count")
plt.title("Simulation of 2 Dice Rolls")

plt.show()
```
